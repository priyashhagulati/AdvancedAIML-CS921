```python
#Task 3: Activation functions
# Linear Act. Function
def lin_fun(x):
  return x

# Step function
def step_fun(x):
  return np.where(x >= 0, 1, 0)

# RELU Act.function
def relu_fun(x):
  return np.maximum(0, x)

def LeakyReLU_fun(x):
  y = np.maximum(0.1*x,x)
  return y

# Sigmoid function
def sigmoid_fun(x):
  return 1 / (1 + np.exp(-x))

x_values = np.linspace(-10, 10, 1000)

y_linear = lin_fun(x_values)
y_step = step_fun(x_values)
y_relu = relu_fun(x_values)
y_leakyrelu = LeakyReLU_fun(x_values)
y_sigmoid = sigmoid_fun(x_values)

# Plot Linear Function
plt.figure(figsize=(8, 5))
plt.plot(x_values, y_linear, label='Linear Function', color='blue')
plt.title('Linear Activation Function')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.axvline(0, color='gray', linestyle='--', linewidth=0.7)
plt.axhline(0, color='gray', linestyle='--', linewidth=0.7)
plt.legend()
plt.show()

# Plot Step Function
plt.figure(figsize=(8, 5))
plt.plot(x_values, y_step, label='Step Function', color='green')
plt.title('Step Activation Function')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.axvline(0, color='gray', linestyle='--', linewidth=0.7)
plt.axhline(0, color='gray', linestyle='--', linewidth=0.7)
plt.legend()
plt.show()

# Plot ReLU Function
plt.figure(figsize=(8, 5))
plt.plot(x_values, y_relu, label='ReLU Function', color='red')
plt.title('ReLU Activation Function')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.axvline(0, color='gray', linestyle='--', linewidth=0.7)
plt.axhline(0, color='gray', linestyle='--', linewidth=0.7)
plt.legend()
plt.show()

# Plot Leaky_ReLU Function
plt.figure(figsize=(8, 5))
plt.plot(x_values, y_leakyrelu, label='Leaky ReLU Function', color='pink')
plt.title('Leaky ReLU Activation Function')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.axvline(0, color='gray', linestyle='--', linewidth=0.7)
plt.axhline(0, color='gray', linestyle='--', linewidth=0.7)
plt.legend()
plt.show()

# Plot Sigmoid Function
plt.figure(figsize=(8, 5))
plt.plot(x_values, y_sigmoid, label='Sigmoid Function', color='purple')
plt.title('Sigmoid Activation Function')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.axvline(0, color='gray', linestyle='--', linewidth=0.7)
plt.axhline(0, color='gray', linestyle='--', linewidth=0.7)
plt.legend()
plt.show()
```

```python
#Task 6: Implementing softmax activation function
input = [1.3, 5.1, 2.2, 0.7, 1.1]

def softmax_fun(input):
  power_e = np.exp(input)
  sum_power_e = np.sum(power_e)
  y=power_e/sum_power_e
  return y

y=softmax_fun(input)
print(y)
```
