```python
#Task 3:
#1. Linear Act. Function
def lin_fun(x):
  y=x
  return y
#2. Step function
#y={1;x>=0}
#y={0;x<0}

#3. RELU Act.function
#y={x;x>0}
#y={0;x<0}

#4. Sigmoid
#log = 1/1+e^-x
#tan = e^x-e^-x/e^x+e^-x
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Linear Act. Function
def lin_fun(x):
  return x

# Step function
def step_fun(x):
  return np.where(x >= 0, 1, 0)

# RELU Act.function
def relu_fun(x):
  return np.maximum(0, x)

# Sigmoid function
def sigmoid_fun(x):
  return 1 / (1 + np.exp(-x))

x_values = np.linspace(-10, 10, 1000)

y_linear = lin_fun(x_values)
y_step = step_fun(x_values)
y_relu = relu_fun(x_values)
y_sigmoid = sigmoid_fun(x_values)

# Plot Linear Function
plt.figure(figsize=(8, 5))
plt.plot(x_values, y_linear, label='Linear Function', color='blue')
plt.title('Linear Activation Function')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.axvline(0, color='gray', linestyle='--', linewidth=0.7)
plt.axhline(0, color='gray', linestyle='--', linewidth=0.7)
plt.legend()
plt.show()

# Plot Step Function
plt.figure(figsize=(8, 5))
plt.plot(x_values, y_step, label='Step Function', color='green')
plt.title('Step Activation Function')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.axvline(0, color='gray', linestyle='--', linewidth=0.7)
plt.axhline(0, color='gray', linestyle='--', linewidth=0.7)
plt.legend()
plt.show()

# Plot ReLU Function
plt.figure(figsize=(8, 5))
plt.plot(x_values, y_relu, label='ReLU Function', color='red')
plt.title('ReLU Activation Function')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.axvline(0, color='gray', linestyle='--', linewidth=0.7)
plt.axhline(0, color='gray', linestyle='--', linewidth=0.7)
plt.legend()
plt.show()

# Plot Sigmoid Function
plt.figure(figsize=(8, 5))
plt.plot(x_values, y_sigmoid, label='Sigmoid Function', color='purple')
plt.title('Sigmoid Activation Function')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.axvline(0, color='gray', linestyle='--', linewidth=0.7)
plt.axhline(0, color='gray', linestyle='--', linewidth=0.7)
plt.legend()
plt.show()
```
